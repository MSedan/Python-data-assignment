```python
component = pd.read_csv('COMPONENT_CODES.csv')
activity = pd.read_csv('ACTIVITY_LOG.csv')
user = pd.read_csv('USER_LOG.csv')
```

```python
print(activity[activity['Component'].isin(['System', 'Folder'])].shape[0])

print(activity[activity['Component'] == 'Folder'].shape[0])
```

```python
activity = activity.rename(columns={'User Full Name *Anonymized': 'USER_ID'})
user = user.rename(columns={'User Full Name *Anonymized': 'USER_ID'})

print(activity)
print(user)
```

```python
acti_comp = pd.merge(activity, component, on='Component', how='left')

print(acti_comp.head(10))
```

```python
final_df = pd.concat([acti_comp, user], axis=1)
```

```python
final_df
```

```python
print(final_df.columns)
```

```python
final_df = final_df.loc[:, ~final_df.columns.duplicated()]
```

```python
final_df
```

```python
pivot_df = final_df.groupby(['USER_ID', 'Component']).size().unstack(fill_value=0)

pivot_df
```

```python
pivot_df.sort_values(by='USER_ID', ascending=True, inplace=True)
pivot_df
```
